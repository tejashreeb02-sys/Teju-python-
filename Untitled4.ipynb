```python
a = int(input("Enter the value of a = "))
b = int(input("Enter the value of b = "))

# Arithmetic Operations

# Addition
c = a + b
print(c)

# Subtraction
d = a - b
print(d)

# Multiplication
e = a * b
print(e)

# Division
f = a / b
print(f)

# Modulus (remainder)
g = a % b
print(g)

# Floor Division
h = a // b
print(h)

# Exponent (power)
i = a ** b
print(i)
```
