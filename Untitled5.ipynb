```python
a = int(input("Enter the value of a = "))
b = int(input("Enter the value of b = "))

# Logical operators binary number system One or Zero, logical high or logical low, True or False

# Logical AND
c = a and b
print(c)

# Logical OR
d = a or b
print(d)

# Logical NOT
e = not a
print(e)

# Logical NOT
f = not b
print(f)
```
