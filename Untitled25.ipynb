```python
# 1. Sort dictionary by values

d = {'a': 3, 'b': 1, 'c': 2}

asc = dict(sorted(d.items(), key=lambda x: x[1]))
desc = dict(sorted(d.items(), key=lambda x: x[1], reverse=True))

print("Ascending:", asc)
print("Descending:", desc)

# 2. Check if key exists in dictionary

d = {'name': 'sakshi', 'age': 20}

key = 'age'

if key in d:
    print("Key exists")
else:
    print("Key not found")

# 3. Merge two dictionaries

d1 = {'a': 1, 'b': 2}
d2 = {'c': 3, 'd': 4}

d1.update(d2)
print(d1)

# 4. Add element to tuple

t = (1, 2, 3)

t = t + (4,)
print(t)

# 5. Tuple with different data types

t = (10, "Hello", 3.5, True)

print(t)


# 6. Sum of list elements

list1 = [10, 20, 30, 40]

print("Sum =", sum(list1))


# 7. Find largest element in list

list1 = [10, 50, 30, 80, 40]

print("Largest =", max(list1))


# 8. Add elements to a set

s = {1, 2, 3}

s.add(4)
s.update([5, 6])

print(s)


# 9. Reverse an array

from array import *

arr = array('i', [1, 2, 3, 4, 5])

arr.reverse()
print(arr)

# 10. Traverse an array

from array import *

arr = array('i', [10, 20, 30, 40, 50])

for i in arr:
    print(i)
```
